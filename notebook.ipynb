```python
import sqlite3
import pandas as pd
from pathlib import Path

# Database path
DB_PATH = Path("data/articles.db")


def get_connection():
    """Get database connection."""
    if not DB_PATH.exists():
        print(f"❌ Database not found at: {DB_PATH}")
        return None
    return sqlite3.connect(DB_PATH)
```
